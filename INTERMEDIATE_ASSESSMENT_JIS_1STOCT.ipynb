# **REGRESSION**

```python
# Features for regression
X_reg = df.drop(columns=["chol", "target"])

# Target variable
y_reg = df["chol"]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = X_reg.select_dtypes(include=["int64", "float64"]).columns

plt.figure(figsize=(14, 6))
sns.boxplot(data=X_reg[numeric_cols])
plt.title("Outliers in Regression Features Before Treatment")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
for col in numeric_cols:
    Q1 = X_reg[col].quantile(0.25)
    Q3 = X_reg[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    X_reg[col] = X_reg[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )
```

```python
plt.figure(figsize=(14, 6))
sns.boxplot(data=X_reg[numeric_cols])
plt.title("Outliers in Regression Features After Treatment")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

print("Training:", X_train_reg.shape, y_train_reg.shape)
print("Testing:", X_test_reg.shape, y_test_reg.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler_reg = StandardScaler()

numeric_features = ["age", "trestbps", "thalach", "oldpeak"]

X_train_reg = X_train_reg.copy()
X_test_reg = X_test_reg.copy()

X_train_reg[numeric_features] = scaler_reg.fit_transform(
    X_train_reg[numeric_features]
)

X_test_reg[numeric_features] = scaler_reg.transform(
    X_test_reg[numeric_features]
)
```

### REGRESSION MODELS

```python

from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor

# Initialize models
lr = LinearRegression()

svr = SVR()

rf_reg = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# Train the models
lr.fit(X_train_reg, y_train_reg)
svr.fit(X_train_reg, y_train_reg)
rf_reg.fit(X_train_reg, y_train_reg)
```

```python
#Making Predictions

# Predictions
y_pred_lr = lr.predict(X_test_reg)
y_pred_svr = svr.predict(X_test_reg)
y_pred_rf_reg = rf_reg.predict(X_test_reg)
```

```python
#Evaluation

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Linear Regression
mae_lr = mean_absolute_error(y_test_reg, y_pred_lr)
mse_lr = mean_squared_error(y_test_reg, y_pred_lr)
r2_lr = r2_score(y_test_reg, y_pred_lr)

# Support Vector Regression
mae_svr = mean_absolute_error(y_test_reg, y_pred_svr)
mse_svr = mean_squared_error(y_test_reg, y_pred_svr)
r2_svr = r2_score(y_test_reg, y_pred_svr)

# Random Forest Regression
mae_rf = mean_absolute_error(y_test_reg, y_pred_rf_reg)
mse_rf = mean_squared_error(y_test_reg, y_pred_rf_reg)
r2_rf = r2_score(y_test_reg, y_pred_rf_reg)

print("Linear Regression:", mae_lr, mse_lr, r2_lr)
print("SVR:", mae_svr, mse_svr, r2_svr)
print("Random Forest:", mae_rf, mse_rf, r2_rf)
```

```python
#Random Forest Regressor has the highest R² and lowest errors among the three models, so it is the best-performing regression model on this test set according to the chosen metric
```

# HYPERPARAMETER TUNING

```python
#Grid Search for Random Forest Classification

from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

rf_param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10],
    "min_samples_split": [2, 5]
}

rf_grid = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid=rf_param_grid,
    scoring="recall",
    cv=2,
    n_jobs=-1
)

rf_grid.fit(X_train_scaled, y_train)

rf_clf_tuned = rf_grid.best_estimator_

print("Best Parameters:", rf_grid.best_params_)
print("Best CV Recall:", rf_grid.best_score_)
```
